```
print(a['Type_of_order'].value_counts())
```

```
type_counts = a['Type_of_order'].value_counts()
plt.figure(figsize=(6,4))
type_counts.plot(kind='bar')
plt.title('Distribution of Type of Order')
plt.xlabel('Type of Order')
plt.ylabel('Count')
plt.savefig('type_counts.png', bbox_inches='tight')
plt.show()
```

```
a['Type_of_order'].value_counts().plot(
    kind='pie',
    autopct='%1.1f%%'
)

plt.title('Percentage of Order Types')
plt.ylabel('')
plt.show()
```

```
print(a['Delivery_person_Ratings'].value_counts())
```

```
a['Delivery_person_Ratings'].value_counts().sort_index().plot(
    kind='bar',
    figsize=(10, 5)
)

plt.title('Distribution of Delivery Person Ratings')
plt.xlabel('Rating')
plt.ylabel('Count')
plt.show()
```

```
# Parse the Order_Date to datetime to extract order distribution by date or year
a['Order_Date'] = pd.to_datetime(a['Order_Date'], format='%d-%m-%Y', errors='coerce')
order_dates = a['Order_Date'].value_counts().sort_index()

# Plot the order counts over time
plt.figure(figsize=(12, 5))
order_dates.plot(kind='line', marker='o', color='purple', linewidth=2)
plt.title('Daily Order Trends')
plt.xlabel('Date')
plt.ylabel('Number of Orders')
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

```
# Extract Year-Month from Order_Date to see order distribution by month
a['Year_Month'] = a['Order_Date'].dt.to_period('M')
monthly_orders = a['Year_Month'].value_counts().sort_index()
print("Order Counts by Month:")
print(monthly_orders)
```

```
city_counts = a['City'].value_counts()

plt.figure(figsize=(10, 5))
city_counts.plot(kind='bar', color='skyblue', edgecolor='black')
plt.title('Distribution of Orders by City Type')
plt.xlabel('City Type')
plt.ylabel('Number of Orders')
plt.xticks(rotation=0)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.savefig('orders_by_city.png', bbox_inches='tight')
plt.show()
```

```
plt.figure(figsize=(10, 5))
monthly_orders.plot(kind='bar', color='orange', edgecolor='black')
plt.title('Monthly Order Trends (2022)')
plt.xlabel('Month')
plt.ylabel('Number of Orders')
plt.xticks(rotation=0)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.savefig('monthly_orders.png', bbox_inches='tight')
plt.show()
```
